# Smart Performance Analytics System

## Projenin Amacı:
Bu projenin amacı Numpy kullanarak çok boyutlu performans verileri üzerinde temel analizler yapabilmektir. Gerçek hayata benzer bir senaryo üzerinden çalışan performanslarının incelenmesi hedeflenmiştir.

## Numpy neden kullanılır?
- Çok boyutlu dizilerle çalışmayı kolaylaştırır.
- Hızlı ve verimli vektörel hesaplama yapmayı sağlar.
- Döngü kullanmadan toplu işlemler gerçekleştirmeyi mümkün kılar.

## Bu projede Kullanılan Numpy konuları:
- NumPy array oluşturma
- Numpy dizi (array) ve matrix yapıları
- İndeksleme ve slicing
- Axis kavramı
- Vektörel (toplu) işlemler
- Temel NumPy metodları


```python
import numpy as np
#8-> çalışan (rows)
#6-> aylar (columns)

#8x6 boyutunda performans matrisi oluştur ve değerler 50-100 arası random olsun
performans = np.random.randint(50, 100, size=(8,6))

performans
```

```python
#3 Array özellikleri (shape, ndim, dtype)
#Toplam eleman sayısı

#Dizi boyutu shape (row, column)
diziBoyutu = performans.shape
print(f"Shpae of array: {diziBoyutu}")

#Dizinin kaç boyutlu olduğu 
kacBoyutlu = performans.ndim
print(f"Number of dimensions: {kacBoyutlu}")

#Dizinin veri tipi
veriTipi = performans.dtype
print(f"Data type of array: {veriTipi}")

#Toplam eleman sayısı
toplamEleman = performans.size
print(f"Total number of elements: {toplamEleman}")
```

```python
#4 Numpy indeksleme

#İlk çalışanın tüm aylarını seç -- çalışanlar satırlardı --  ilk çalışan 0. indekstir
ilkCalisan = performans[0] # -> ilk calisan
print(f"İlk çalışanın tüm ayları: {ilkCalisan}")

#Son ayın tüm çalışanlarını seç -- aylar sütunlardı -- 
sonAy = performans[:, -1] #[bütün satırlar, son sütun] yani son sütundaki bütün satırlardaki çalışanları aldık
print(f"Son ayın tüm çalışanları: {sonAy}")

#İlk 3 çalışanın ilk 2 ayını seç
ucCalisanIkiAy = performans[:3, :2] # ilk 3 çalışan 0,1,2. indekslerdir ve ilk 2 ay da 0,1. indekslerdir
print(f"İlk 3 çalışanın ilk 2 ayı:\n {ucCalisanIkiAy}")
```

```python
#5 Matrix indeksleme

#Belirli bir çalışanın belirli ay skorunu al
#6. çalışanın, 4. ay performansını alalım--> 6. çalışan 5. indekstir, 4. ay da 3. indekstir
belirliCalisan = performans[5, 3]
print(f"Altıncı çalışanın 4. ay performansı: {belirliCalisan}")

#Bir alt matris (sub-matrix oluştur)
#2. ve 3. çalışanların, 4. ve 5. ay skorları
sub_matrix = performans[1:3, 3:5]
print(f"2. ve 3. çalışanların dördüncü ve beşinci ay skorları:\n {sub_matrix}")
```

```python
#6 Numpy Operasyonlar
# Tüm skorlara +5 puan ekle
degismisSkorlar = performans + 5
print(f"Tüm skorlara 5 eklenmiş hali:\n {degismisSkorlar}")
print(" ")

#Skorları 100 ile normalize et (0-1) e çek yani
normalize = performans / 100
print(f"Normalize edilmiş skorlar:\n {normalize}")
print(" ")

#İki farklı ayın skorlarını karşılaştır (2. ay vs 5. ay)
ikiAyKarsilastirma = performans[:,5] > performans[:, 1]
print(f"Beşinci ayın skoru ikinci aydan yüksek mi?: {ikiAyKarsilastirma}")
```

```python
# 7 Numpy metotları

#çalışan bazlı ortalama (axis=1) -> satır boyunca işlem yapar
calisanOrtalama = np.mean(performans, axis=1)
print(f"Çalışan bazlı ortalama performansı:\n {calisanOrtalama}")
print(" ")

#Ay bazlı ortalama (axis=0)
ayOrtalama = np.mean(performans, axis=0)
print(f"Ay bazlı ortalama performansı: {ayOrtalama}")

#En yüksek performans
enYuksekPerf = np.max(performans)
print(f"En yüksek performans: {enYuksekPerf}")

#En düşük performans 
enDusukPerf = np.min(performans)
print(f"En düşük performans: {enDusukPerf}")
```

```python
#8 Numpy dizi metotları
# reshape - 8x6 = 48 di, 4x12 yapabiliriz yani 4 satır 12 sütun
reshapedData = performans.reshape(4,12)
print(f"Reshape yapılmış veri (4x12):\n {reshapedData}")
print(" ")

#transpose -> satır ve sütunlar yer değiştirir satırlar ayları, sütunlar çalışanları temsil eder
transposed = performans.T
print(f"Transpose edilmiş hali:\n {transposed}")
print(" ")

#flatten 2D'den 1D'ye çevirir yani düzleştirir
flattenData = performans.flatten()
print(f"Flatten edilmiş ve 1D hale getirilmiş hali:\n {flattenData}")
```

```python
# 9 Genel performans özeti
# calisanOrtalama'yı yukarıda aldık
#En başarılı çalışan 
enBasariliCalisan = np.argmax(calisanOrtalama) #indeks veriyor
print(f"En başarılı çalışan: Çalışan {enBasariliCalisan+1}") #indeksi + 1 yaptık

#En yüksek ay ortalaması
#aybazlıortalamayı yukarıda aldık yine
enYuksekAy = np.argmax(ayOrtalama)
print(f"En yüksek ay ortalaması: Ay {enYuksekAy+1}")

#Genel ortalama performansı
genelOrtalamaPerf = np.mean(performans)
print(f"Genel ortalama performans: {genelOrtalamaPerf:.2f}")
```
