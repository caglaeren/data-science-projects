# Smart Customer Data Analysis System 

## Projenin Amacı:
Pandas kütüphanesinin temel yapı taşlarını kullanarak gerçek hayatta karşılığı olan müşteri harcama verilerini analiz etmektir.

## Pandas Neden Kullanılır?
- Tablo bazlı verilerle çalışmayı kolaylaştırır.
- Veriye anlamlı indeksler atamayı sağlar.
- Gerçek hayattaki müşteri, satış ve finans verilerini analiz etmek için idealdir.

# Bu Projede Kullanılan Pandas Konuları:
- Pandas giriş
- Series
- Series özellikleri
- DataFrame
- DataFrame indeksler
- İndeks değiştirme
- Multi-index

```python
import pandas as pd

#2 Aylık harcamaları temsil eden bir series oluştur ve indeks olarak ay isimlerini kullan
harcananPara = [2000, 4050, 3300, 5060, 9195, 8200]
aylar = ["Ocak", "Şubat", "Mart", "Nisan", "Mayıs", "Haziran"] #6 ay öğrenmesi için yeterlidir
monthlySpending = pd.Series(data=harcananPara, index=aylar, name="Aylık harcamalar")
monthlySpending
```

```python
#3 Series Özellikleri
# values, index, dtype, ortalama harcama

#Seri içindeki değerler
seriesValues = monthlySpending.values
print(f"Seri içindeki değerler: {seriesValues}")

#index bilgisi
indexSeries = monthlySpending.index
print(f"Seri içindeki indeksler: {indexSeries}")

#veri tipi
dataTypeSeries = monthlySpending.dtype
print(f"Seri içindeki veri tipi: {dataTypeSeries}")

#ortalama harcama
meanSpend = monthlySpending.mean()
print(f"Ortalama harcama: {meanSpend:.3f} TL.")
```

```python
#4 DataFrame oluştur
#Sütunlar -> CustomerName, City, MonthlySpending, MembershipType (Amaç: Tablolu veri mantığı)
customerDataFrame = pd.DataFrame({
    "CustomerName": ["Ali", "Merve", "Suzan", "Kıvanc", "Canan"],
    "City": ["Ankara", "Bursa", "Ankara", "İstanbul", "Rize" ],
    "MonthlySpending": [5400, 2800, 6400, 7105, 8910],
    "MembershipType": ["Silver", "Platinum", "Gold", "Silver", "Bronze"]
    
})

customerDataFrame
```

```python
#5 DataFrame Indeksleme

#Bir sütunu indeks olarak ayarla
columnIndexed = customerDataFrame.set_index("CustomerName")
columnIndexed

#indeks üzerinden satır seçme
columnIndexed.loc["Kıvanc"]

```

```python
#6 Index Değiştirme

#Indexi sıfırla (reset_index)
resetIndex = columnIndexed.reset_index()
resetIndex

#Yeni bir indeks tanımla (set_index)
newIndex = resetIndex.set_index("MembershipType")
newIndex
```

```python
#7 Multi-Index oluşturma
multiIndex = customerDataFrame.set_index(["City", "MembershipType"])
multiIndex

#Belli bir şehir-üyelik tipi seç
multiIndex.loc[("Ankara","Silver")]
```

```python
#Son çıktı
#Toplam müşteri sayısı:
totalCustomers = len(customerDataFrame) #kaç satır yani kaç müşteri olduğunu veriyor

#En yüksek aylık harcama
maxMonthlySpending = customerDataFrame["MonthlySpending"].max() #max değeri bulacak

#En çok müşteri olan şehir
topCity = customerDataFrame["City"].value_counts().idxmax() #value_counts() bir sütundaki her bir unique değerin kaç kez tekrarladığını hesaplar ve büyükten küçüğe döndürür
#max() yazınca 2 döndürüyor ama idxmax() yazınca Ankara döndürüyor.
#idxmax(): index of maximumdur ve en büyük değer kimde sorusunun cevabıdır

print(f"Toplam müşteri sayısı: {totalCustomers}")
print(f"En yüksek aylık harcama: {maxMonthlySpending} TL.")
print(f"En çok müşteri olan şehir: {topCity}")
```
