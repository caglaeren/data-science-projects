### 2- Veriyi üret

```python
import numpy as np
import pandas as pd
from data_generator import generate_sales_data
from preprocessing import preprocessing_data
from analytics import city_sale, product_sale, monthly_trend
from reporting import decision_report

#Veriyi üret
new_df = generate_sales_data(size=1000) #ham verimiz
new_df.head(20) #ilk 20 tanesini verir
```

### 3- Veriyi temizle

```python
clean_df = preprocessing_data(new_df) #preprocessing fonksiyonumuzda temizleme yapmıştık
clean_df.head(20)
```

### 4- Analiz et

```python
#analytics.py dosyamızdan alacağız
city_analysis = city_sale(clean_df) #şehir bazlı sum/mean/count
product_analysis = product_sale(clean_df) #ürün bazlı
monthly_analysis = monthly_trend(clean_df) #ay bazlı
```

### 5-Sonuçları Yazdır

```python
#raporu alalım (sehir ve ürün)
rapor = decision_report(city_report = city_analysis, product_report=product_analysis) #best/worst city, best product, has risk getirir

#şehirleri bölgeye yazalım:
city_to_region= {
    "İstanbul" : "Marmara",
    "Ankara" : "Ic Anadolu",
    "Mersin" : "Akdeniz",
    "Rize" : "Karadeniz",
    "Van" : "Dogu Anadolu",
    "Aydın" : "Ege",
    "Konya": "Ic Anadolu" , 
    "Samsun": "Karadeniz",
    "Denizli" : "Ege"
}

#En iyi ay
best_month = monthly_analysis["sum"].idxmax()

#Riskli bölge (en kötü şehirden)
worst_c = rapor["Worst City"][0] #şehir, değer
risk_region = city_to_region.get(worst_c, "Unknown")

#Toplam satış
total_sale = clean_df["Amount"].sum()

```

### 6- Excel çıktısını al

```python
file = "final_sales_report.xlsx"

with pd.ExcelWriter(file) as writer:
    #temiz veri
    clean_df.to_excel(writer, sheet_name = "Clean_Data", index=False)

    #analiz tabloları
    city_analysis.to_excel(writer, sheet_name="City_Analysis")
    product_analysis.to_excel(writer, sheet_name="Products_Analysis")
    monthly_analysis.to_excel(writer, sheet_name="Monthly_Analysis")

#Son çıktı
print(f"Toplam satış: {total_sale} TL.")
print(f"En karlı şehir: {rapor['Best City'][0]}")
print(f"En iyi ay: {best_month}")
print(f"Riskli bölge: {risk_region} Bölgesi")
print(f"Excel raporu oluşturuldu")
```
