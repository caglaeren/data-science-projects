```python
#2- Numpy import ve veri oluşturma
import numpy as np

# 7 gün x 5 ürünlük satış verisi
sales = np.random.randint(5, 30, size = (7,5)) #5-30 arası rastgele satış adetleri, 7 gün ve 5 ürün (7 satır 5 sütun)
```

```python
sales
```

```python
# 3- Array özellikleri
# shape, ndim, dtype ve toplam eleman sayısı

#Array boyutu - shape (satır, sütun)
sales_boyutu = sales.shape
print(f"Shape: {sales_boyutu}")

#Arrayın kaç boyutlu olduğu ndim
arrayinBoyutlu = sales.ndim
print(f"Number of dimensions (Dizinin boyutu): {arrayinBoyutlu}")

#Array içindeki veri tipi - dtype
arrayVeritipi = sales.dtype
print(f"Data type of array: {arrayVeritipi}")

#Toplam eleman sayısı - size
toplamEleman = sales.size
print(f"Total number of elements: {toplamEleman}")
```

```python
#4- Indexing and Slicing

#Belirli bir günün satışlarını al (5.günü alalım)
belirliGunSatisi = sales[4]
print(f"Beşinci günün satışları: {belirliGunSatisi}")

#Belirli bir ürünün haftalık satışını al - 2. ürün olsun -> index 1
belirliHaftalikSatis = sales[:, 1] # [satır, sütun] -> [günler, ürünler] yani [hepsini al, 2. ürünü al (indeksi 1)]
print(f"2. ürünün haftalık satışları: {belirliHaftalikSatis}")

#ilk 3 günün satışları
ilkUcGununSatis  = sales[:3, :] # -> [:3] demek 0,1,2. indeksleri al demek yani ilk 3 gün, [, :] ise tüm ürünleri al demek
print(f"İlk 3 günün satışları:\n {ilkUcGununSatis}")
```

```python
#5- Temel Numpy metotları

#Toplam satış (sum)
toplamSatis = sales.sum()

#Günlük ortalama satış (mean) - gün bazlı yani satır bazlı
gunlukOrtalamaSatis = sales.mean(axis=1) # axis=1 -> satır boyunca yani gün

#En yüksek satış - max
enYuksekSatis = sales.max()

#En düşük satış - min
enDusukSatis = sales.min()


print(f"Toplam satış: {toplamSatis}")
print(f"Günlük ortalama satış: {gunlukOrtalamaSatis}")
print(f"En yüksek satış: {enYuksekSatis}")
print(f"En düşük satış: {enDusukSatis}")
```

```python
# 6- Axis kullanımı

#gün bazlı toplam satış
gunTopSatis = sales.sum(axis=1) #gün bazlı yani satır bazlı toplam satış
print(f"Gün bazlı toplam satış: {gunTopSatis}")

#ürün bazlı toplam satış
urunTopSatis = sales.sum(axis=0) #ürün bazlı toplam satış
print(f"Urun bazlı toplam satış: {urunTopSatis}")

#axis = 1-> satır boyunca işlem yapar yani bizim problemimizde günlere denk gelir (Yani sütunlar boyunca soldan sağa toplar )
#axis= 0 -> sütun boyunca işlem uyapar bizim problemimizde ürünlere denk gelir (Yani aslında satırlar boyunca yukarıdan aşağıya toplar)
```

```python
#7- Numpy dizi metotları
#reshape -> Arrayin şeklini değiştirir bizim verimiz 7 x 5 = 35 elemandı
# 35 eleman korunarak farklı boyuta sokulabilir
reshaped_sale = sales.reshape(5,7) #eleman sayısı aynı sadece şekil değişir
print(f"Reshape 5x7:\n {reshaped_sale}")

print(" ")

# transpose -> satır ve sütunları yer değiştirir yani günler ve ürünler yer değiştirir
transposed = sales.T
print(f"Transposed data:\n {transposed}")
print(" ")

#flatten -> 2d boyutlu diziyi tek boyuta (vektörel) çevirir, kopya oluşturur
flattened = sales.flatten()
print(f"Flatten data:\n {flattened}")

#ravel -> kopya oluşturmadan 1D array döner
#raveled = sale.ravel()
```

```python
#7- Çıktılar

#Toplam haftalık satış (tüm gün + tüm ürünler)
print(f"Toplam haftalık satış: {toplamSatis}")

#En çok satan ürün (ürün bazlı toplamdan)
#np.argmax() -> bir array içindeki en büyük değerin indeksini verir
enCokSatanUrun = sales.sum(axis=0) #axis=0 ürünlerdi
enCokSatanUrunIndex = np.argmax(enCokSatanUrun) #en cok satan ürünün indeksini aldık
print(f"En çok satan ürün: Ürün {enCokSatanUrunIndex + 1}") #index 0->ürün 1, biz saymaya 1 den başladığımız için 0+1=ürün1 olacak

#En iyi gün
enIyiGun = sales.sum(axis=1) #axis1 = günler
enIyiGununIndex = np.argmax(enIyiGun)
print(f"En iyi gün: Gün {enIyiGununIndex + 1}") # mesela 5. günün indeksi 4 burada index + 1 diyerek 5. gün olarak alıyoruz
```
