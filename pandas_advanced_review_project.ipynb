Pandas veri analizi için güçlü ve esnek araçlar sunar.

## Kapsanan Pandas Konuları:

- Pandas ileri operasyonlar
- Excel ile çalışma (`read_excel`, `to_excel`)

### 2-Veri Setlerini Oluşturma

```python
#Üç ayrı DataFrame oluştur
import pandas as pd
import numpy as np

#1-Customers DF
customers = pd.DataFrame({
    "CustomerID": [99, 100,102,103,104,105,206,207,208,210],
    "CustomerName":["Doja","Katy","Cardi", "Wayne", "Jennie","Fouzia","Jony","Ariana" ,"Abel","Rihanna"],
    "City": ["Los Angeles","London","İstanbul","İstanbul","Seoul","Paris","Moscow","New York", "Los Angeles","London"],
    "Age" : [30, 41, 33, np.nan, 30, 25, 28, 32, np.nan, 38] #bazı yaşlar eksik
})

#2-Sales DF
sales = pd.DataFrame({
    "SaleID": [1,2,3,4,5,6,7,8],
    "CustomerID": [99, 100,102,103,104,105,206,207],
    "Product": ["Laptop","Kulaklık","Telefon","Kulaklık", "Tablet", "Laptop","Telefon","Saç maşası"],
    "Amount": [45000, 2300, np.nan, np.nan, 12860, 72950, 37000, 3100] #bazı değerler nan
})

#3-Regions DF
regions = pd.DataFrame({
    "City": ["Los Angeles", "London", "İstanbul","Seoul","Paris","Moscow","New York"],
    "Region": ["West America", "Europe","Asia","Asia", "Europe", "Asia","East America"],
})

print("CUSTOMERS")
display(customers)
print("---------------------------------------------------------- ")
print(" ")

print("SALES")
display(sales)
print("---------------------------------------------------------- ")
print(" ")

print("REGIONS")
display(regions)
```

### 3-Eksik Veriler

```python
#Eksik değerleri tespit edelim
#customers icin
customersEksikDeger = customers.isna().sum() 
print(f"Customers Eksik değerleri: {customersEksikDeger}")
print(" ")
#sales icin
salesEksikDeger = sales.isna().sum()
print(f"Sales Eksik değerleri: {salesEksikDeger}")
print(" ")
#Yaş için ortalama (mean) ile doldur
yasOrtalama = customers["Age"].mean() #yaş ortalamasını aldık
customers["Age"] = customers["Age"].fillna(yasOrtalama) #yasOrtalamadan gelen sonuç ile doldurduk ve yaşlara eşitledik
print("Yaş ortalaması ile customers değerleri dolduruldu.")
display(customers)

#Satış tutarı eksik olan kayıtları sil
salesClean = sales.dropna(subset=["Amount"])
print("Temizlenmiş satış verisi")
display(salesClean)
```

### 4-Merge işlemi

```python
# customers + sales -> customerID üzerinden birleştir
# ortaya çıkan tabloyu regions ile birleştir
#left-> customers, right->salesClean
customersSales = pd.merge(customers, salesClean, how="inner", on="CustomerID") #sadece satış yapan müşteriler kalsın dediğimiz için inner

#regions ekle -> City üzerinden ekleriz
tamBirlesmisHal = pd.merge(customersSales, regions, how="left", on="City") #tüm müşteriler kalsın, region varsa eklensin

display(tamBirlesmisHal)
```

### 5-Groupby ile Gruplandırma

```python
#Şehir bazlı toplam satış
#agg()-> aggretion yani özetlemedir. veriyi gruplayıp özet hesaplar yani toplam, ortalama, sayım, min/max vb.
citySale = tamBirlesmisHal.groupby("City").agg(ToplamSatis = ("Amount","sum")) #Şehire göre gruplar, her grup için amount sütununu toplar
print("Şehir bazlı toplam satış")
display(citySale)
print(" ")

#Bölge bazlı ortalama satış
regionSale = tamBirlesmisHal.groupby("Region").agg(OrtalamaSatis = ("Amount", "mean"))
print("Bölge bazlı ortalama satış")
display(regionSale)
print(" ")

#Ürün bazlı satış sayısı
productSale = tamBirlesmisHal.groupby("Product").agg(UrunSatis = ("Amount","count"))
print("Ürün bazlı satış sayısı:")
display(productSale)
```

### 6-Concat İşlemleri

```python
#Farklı aylara ait iki satış tablosu oluştur. Satır bazlı birleştir(axis=0)
salesMart = pd.DataFrame({
    "SaleID": [1,5,7,8],
    "CustomerID": [99,104,206,207],
    "Product": ["Laptop","Tablet","Telefon","Saç maşası"],
    "Amount": [45000, 12860, 37000, 3100],
    "Month": ["2026-03", "2026-03", "2026-03", "2026-03"]
})

salesEyl = pd.DataFrame({
    "SaleID": [2,6,7],
    "CustomerID": [100,105,207],
    "Product": ["Kulaklık","Laptop","Saç maşası"],
    "Amount": [2300,72950, 3100],
    "Month": ["2026-09", "2026-09", "2026-09"]
    
})

#Satır bazlı birleştirelim
salesAllMonths = pd.concat([salesMart,salesEyl], axis=0, ignore_index=True)
print("Mart satışları")
display(salesMart)
print(" ")

print("Eylül satışları")
display(salesEyl)
print(" ")

print("Birleştirilmiş aylık satışlar")
display(salesAllMonths)
```

### 7-Pandas ile İleri Operasyonlar

```python
#Koşullu filtreleme
yuksekSatis = salesAllMonths[salesAllMonths["Amount"] > 20000]
print("20 bin üstü fiyata sahip olan satışlar: ")
display(yuksekSatis)
print(" ")

#Yeni sütun oluşturma (Örn: KDV eklenmiş satır) %20 kdv ekleyelim
salesAllMonths["KDV"] = salesAllMonths["Amount"] * 1.20
print("KDV eklenmiş satış tutarı: ")
display(salesAllMonths)

#apply() kullanımı -> Satış tutarına göre kategori oluşturalım
def satisKategori(amount):
    if amount >= 40000:
        return "Premium"
    elif amount >= 25000:
        return "Gold"
    elif amount >= 10000:
        return "Standard"
    else:
        return "Basic"

salesAllMonths["Category"] = salesAllMonths["Amount"].apply(satisKategori)
print("Kateogri eklenmiş hali: ")
display(salesAllMonths)
```

### 8- Excel ile Çalışma

```python
# Temizlenmiş ve analiz edilmiş tabloyu excel'e yaz
salesAllMonths.to_excel("final_sales_report.xlsx", index=False) #dosyayı oluşturdul
print("Excel dosyası yazıldı.")

#Excel dosyasını okuyalım
salesFromExcel = pd.read_excel("final_sales_report.xlsx") #dosyayı tekrar yükledik 
print("Excel'den okunan veri")
display(salesFromExcel)
```
